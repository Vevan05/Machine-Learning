```python
import numpy as np
import tensorflow as tf
import pandas as pd
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
import matplotlib.pyplot as plt
from urllib.request import urlopen
```

```python
max_features = 10000
maxlen = 500
batch_size = 32
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=max_features)
print(len(X_train), "train sequences")
print(len(X_test), "test sequences")
```

```python
X_train = sequence.pad_sequences(X_train, maxlen=maxlen)
X_test = sequence.pad_sequences(X_test, maxlen=maxlen)
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
data = pd.read_csv(urlopen("https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"))
values = data["Passengers"].values.reshape(-1,1)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense

model_gru = Sequential([
    tf.keras.layers.Input(shape=(maxlen,)),
    tf.keras.layers.Embedding(max_features, 128),
    GRU(64, return_sequences=False),
    Dense(1, activation='sigmoid')
])
```

```python
model_gru.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_gru = model_gru.fit(X_train, y_train, epochs=5, batch_size=batch_size, validation_split=0.2)

gru_loss, gru_accuracy = model_gru.evaluate(X_test, y_test)
print(f'GRU Test Accuracy: {gru_accuracy * 100:.2f}%')
```

```python
y_pred_gru = model_gru.predict(X_test)
y_pred_gru_classes = (y_pred_gru > 0.5).astype("int32")
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred_gru_classes)
print("GRU Accuracy:", accuracy)
```
