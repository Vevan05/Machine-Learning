```python
import nltk
from nltk.corpus import brown
from gensim.models import Word2Vec

nltk.download('brown')

sentences = brown.sents(categories='news')
sentences = [[word.lower() for word in sent] for sent in sentences]
```

```python
# CBOW
model1 = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=5,
    sg=0,
    workers=4
)

# Skip-gram
model2 = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=5,
    sg=1,
    workers=4
)
```

```python
print(model1.wv.most_similar('government'))
print(model1.wv['government'].shape)
```

```python
print(model2.wv.most_similar('government'))
print(model2.wv['government'].shape)
```
