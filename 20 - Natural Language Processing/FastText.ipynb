```python
import nltk
nltk.download('brown')

from nltk.corpus import brown

sentences = brown.sents()
sentences = [[word.lower() for word in sent] for sent in sentences]

print("Number of sentences:", len(sentences))
print("Sample sentence:", sentences[0])
```

```python
from gensim.models import FastText

model = FastText(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=5,
    workers=4,
    sg=1
)
```

```python
print(model.wv['government'])
```

```python
print(model.wv.most_similar('government'))
```

```python
print(model.wv['governments'])
```
