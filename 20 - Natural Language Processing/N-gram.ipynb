```python
import nltk

from nltk.corpus import brown
from nltk.corpus import stopwords

nltk.download('stopwords')
nltk.download('brown')
```

```python
import string

tokens = brown.words(categories = 'news')
tokens = [t.lower() for t in tokens]
tokens = [t for t in tokens if t not in string.punctuation]

print(tokens)
```

```python
from nltk import ngrams

N = 2
bigrams = (list(ngrams(tokens, N)))

print(bigrams)
```

```python
freq = {}

for bigram in bigrams:
  if bigram not in freq:
    freq[bigram] = 1
  else:
    freq[bigram] += 1

res = {}
sup = 0

# Replacing words with frequency < 5 with "UNK"
for k, v in freq.items():
  if v >= 5:
    res[k] = v
  else:
    sup += v

res["UNK"] = sup
```

```python
total = len(bigrams)
print(total)

V = 0
unique = {}
for i in tokens:
  if i not in unique:
    V += 1

  unique[i] = 1

print(V)
```

```python
# Laplace add-one Smoothing
probs = {}

for bg, c in freq.items():
  probs[bg] = (freq[bg] + 1)/(total + V)
```

```python
def predict_next(curr):
  candidates = {w2 : prob for (w1, w2), prob in probs.items() if w1 == curr}

  if not candidates: return None
  return max(candidates, key=candidates.get)
```

```python
word = input("Enter word: ")
print(predict_next(word))
```

```python
import random

def sample_next_word(current_word):
    words = []
    p = []

    for (w1, w2), prob in probs.items():
        if w1 == current_word:
            words.append(w2)
            p.append(prob)

    if not words:
        return None

    return random.choices(words, p)[0]
```

```python
def generate_sentence(start_word = "<START>", end_word = "<END>", max_len = 15):
  word = start_word.lower()
  sentence = []

  for _ in range(max_len):
    next_word = sample_next_word(word)
    if next_word is None or next_word == end_word:
      break
    sentence.append(next_word)
    word = next_word

  return " ".join(sentence)
```

```python
print(generate_sentence("center"))
```
